**Question 1:** How many customers are in the file?
(as help, we have added some comments and starter code to help you structure the solution)

```python
from os.path import exists
import csv

# Check if the customers file exists
datafile = "./sample_data/customers.csv"

if not exists(datafile):
    raise SystemExit("Please download the dataset files first!")

# Read the CSV and count customer records
with open(datafile, "r", encoding="utf-8-sig", newline="") as fl:
    csvreader = csv.reader(fl, delimiter=",")

    next(csvreader)  # Skip the header row

    total_customers = 0

    for row in csvreader:
        if row and any(value.strip() for value in row):
            total_customers += 1

print("Total customers:", total_customers)
```

**Question 2:** In how many different states do the customers live in?

```python
import csv

datafile = "./sample_data/customers.csv"

# Store unique states
states = set()

with open(datafile, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    for customer in reader:
        state = customer["State"].strip().upper()

        if state:
            states.add(state)

print("Different states:", len(states))
print("States:", sorted(states))
```

**Question 3** What is the state with most customers?

```python
import csv
from collections import Counter

datafile = "./sample_data/customers.csv"

# Count customers in each state
state_counts = Counter()

with open(datafile, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    for customer in reader:
        state = customer["State"].strip().upper()

        if state:
            state_counts[state] += 1

# Find the state with the most customers
most_common_state, customer_count = state_counts.most_common(1)[0]

print("State with most customers:", most_common_state)
print("Number of customers:", customer_count)
```

**Question 4** What is the state with the least customers?

```python
import csv
from collections import Counter

datafile = "./sample_data/customers.csv"

# Count customers per state
state_counts = Counter()

with open(datafile, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    for customer in reader:
        state = customer["State"].strip().upper()

        if state:
            state_counts[state] += 1

# Find the minimum number of customers
min_customers = min(state_counts.values())

# Include all states tied for the minimum
least_common_states = sorted(
    state for state, count in state_counts.items()
    if count == min_customers
)

print("States with the fewest customers:", least_common_states)
print("Number of customers per state:", min_customers)
```

**Question 5:** What is the most common last name?

```python
import csv
from collections import Counter

datafile = "./sample_data/customers.csv"

# Count occurrences of each last name
last_name_counts = Counter()

with open(datafile, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    for customer in reader:
        last_name = customer["LastName"].strip().casefold()

        if last_name:
            last_name_counts[last_name] += 1

# Find the most common last name
most_common_name, count = last_name_counts.most_common(1)[0]

print("Most common last name:", most_common_name.title())
print("Number of occurrences:", count)
```

The second sample files contains orders placed by customers from the first file. Be careful, this file has many rows and you most likely should not print the contents of the whole file.

**Question #1:** How many unique orders are in the orders.csv file?

**Question #2:** What is the average number of items per order (rounded to two decimal places)?

**Question #3:** What is the highest number of items per order?

**Question #4:** What is the number of orders placed in October 2021?

**Question #5:** Which customer spent the most amount of money in 2021?

**Question #6:** Historically, what is the best month for sales?

```python
import csv
from calendar import month_name
from collections import defaultdict
from datetime import datetime
from decimal import Decimal


# Data loading and preprocessing

# Use Price to calculate product sales.
# Keep OrderTotal for comparison, as the values may differ.

datafile = "./sample_data/orders.csv"

orders_by_id = {}
total_rows = 0
inconsistent_orders = set()

with open(datafile, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    for row in reader:
        order_id = row["OrderID"].strip()

        if not order_id:
            continue

        total_rows += 1

        customer_id = row["CustomerID"].strip()
        order_date = row["Date"].strip()
        order_total = Decimal(row["OrderTotal"].strip())
        product_price = Decimal(row["Price"].strip())

        if order_id not in orders_by_id:
            orders_by_id[order_id] = {
                "customer_id": customer_id,
                "date": order_date,
                "total": order_total,
                "product_total": Decimal("0"),
                "items": 0,
            }
        else:
            # Ensure repeated order records are consistent
            existing = orders_by_id[order_id]

            if (
                existing["customer_id"] != customer_id
                or existing["date"] != order_date
                or existing["total"] != order_total
            ):
                inconsistent_orders.add(order_id)

        # Each row represents one product in the order
        orders_by_id[order_id]["items"] += 1
        orders_by_id[order_id]["product_total"] += product_price

# Stop processing if an order contains conflicting data
if inconsistent_orders:
    raise ValueError(
        f"Found {len(inconsistent_orders)} orders with inconsistent data."
    )


# Question 1: Unique orders

total_orders = len(orders_by_id)

print("Q1 - Unique orders:", total_orders)


# Question 2: Average items per order

total_items = sum(
    order["items"] for order in orders_by_id.values()
)

average_items = round(total_items / total_orders, 2)

print("Q2 - Average items per order:", average_items)


# Question 3: Highest number of items per order

max_items = max(
    order["items"] for order in orders_by_id.values()
)

orders_with_max_items = sorted(
    order_id
    for order_id, order in orders_by_id.items()
    if order["items"] == max_items
)

print("Q3 - Highest items per order:", max_items)
print("     Order IDs:", orders_with_max_items)


# Date validation and sales aggregation

october_2021_orders = 0
orders_in_2021 = 0
orders_without_date = 0
invalid_dates = 0

customer_spending = defaultdict(lambda: Decimal("0"))
customer_order_totals = defaultdict(lambda: Decimal("0"))
monthly_sales = defaultdict(lambda: Decimal("0"))
monthly_order_totals = defaultdict(lambda: Decimal("0"))
period_sales = defaultdict(lambda: Decimal("0"))

# Process each unique order only once
for order in orders_by_id.values():
    order_date = order["date"]

    if not order_date:
        orders_without_date += 1
        continue

    try:
        date = datetime.fromisoformat(order_date)
    except ValueError:
        invalid_dates += 1
        continue

    order_total = order["total"]
    product_total = order["product_total"]

    # Aggregate orders and customer spending for 2021
    if date.year == 2021:
        orders_in_2021 += 1

        if date.month == 10:
            october_2021_orders += 1

        if order["customer_id"]:
            customer_id = order["customer_id"]

            customer_spending[customer_id] += product_total
            customer_order_totals[customer_id] += order_total

    # Aggregate historical product sales by calendar month
    monthly_sales[date.month] += product_total

    # Keep order totals for comparison
    monthly_order_totals[date.month] += order_total

    # Track individual year-month sales
    period = date.strftime("%Y-%m")
    period_sales[period] += product_total


# Question 4: Orders placed in October 2021

print("Q4 - Orders in October 2021:", october_2021_orders)


# Question 5: Highest-spending customer in 2021

top_customer, highest_spending = max(
    customer_spending.items(),
    key=lambda item: item[1]
)

print("Q5 - Top customer ID:", top_customer)
print("     Product spending in 2021:", highest_spending)
print("     Order totals for comparison:",
      customer_order_totals[top_customer])


# Question 6: Best historical month for sales

best_month = max(monthly_sales, key=monthly_sales.get)
best_period = max(period_sales, key=period_sales.get)

print("Q6 - Best calendar month:", month_name[best_month])
print("     Product sales:", monthly_sales[best_month])
print("     Order totals for comparison:",
      monthly_order_totals[best_month])

print("     Best individual year-month:", best_period)
print("     Product sales in that period:", period_sales[best_period])


# Data quality checks

print("\n--- Data Quality Checks ---")
print("Total records:", total_rows)
print("Unique orders:", total_orders)
print("Orders processed in 2021:", orders_in_2021)
print("Orders without date:", orders_without_date)
print("Invalid dates:", invalid_dates)
print("Orders with inconsistent data:", len(inconsistent_orders))
```
